```python
# 1. Import libraries
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
# 2. Set image size & batch size
img_size = 128   # you can change (64,128,224)
batch_size = 32
```

```python
# 3. Data Preprocessing (Train & Test Generators)
train_datagen = ImageDataGenerator(rescale=1./255, validation_split=0.2)
```

```python
train_generator = train_datagen.flow_from_directory("/content/drive/MyDrive/images of DC",target_size=(img_size, img_size),batch_size=batch_size,class_mode="binary",)

val_generator = train_datagen.flow_from_directory("/content/drive/MyDrive/images of DC", target_size=(img_size, img_size),batch_size=batch_size,class_mode="binary",subset="validation")
```

```python
# 4. Build CNN model
model = Sequential([
    Conv2D(32, (3,3), activation='relu', input_shape=(img_size, img_size, 3)),
    MaxPooling2D(pool_size=(2,2)),

    Conv2D(64, (3,3), activation='relu'),
    MaxPooling2D(pool_size=(2,2)),

    Conv2D(128, (3,3), activation='relu'),
    MaxPooling2D(pool_size=(2,2)),

    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.3),
    Dense(64, activation='relu'),
    Dropout(0.3),

    Dense(1, activation='sigmoid')  # binary classification
])
```

```python
# 5. Compile model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
# 6. Train model
history = model.fit(
    train_generator,
    validation_data=val_generator,
    epochs=10)
```

```python
# 7. Save model
model.save("cnn_image_classifier.h5")
```

```python
# 8. Evaluate on validation data
loss, acc = model.evaluate(val_generator)
print(f"✅ Validation Accuracy: {acc:.4f}")
```

```python
from tensorflow.keras.preprocessing import image
import numpy as np

# Load image
img_path = "/content/drive/MyDrive/images of DC/dogs_set/dog.4002.jpg"
img = image.load_img(img_path, target_size=(img_size, img_size))
img_array = image.img_to_array(img) / 255.0
img_array = np.expand_dims(img_array, axis=0)

# Prediction
prediction = model.predict(img_array)

if prediction[0][0] > 0.5:
    print("Predicted: Dog 🐶")
else:
    print("Predicted: Cat 🐱")
```
